```python
import pandas as pd
df = pd.read_csv('tabular_wayne_before.csv')
def process_sex_column(df):
    # 删除'Sex'列中的'Others'
    df = df[df['Sex'] != 'Other']
    
    # 将'Female'替换为0,'Male'替换为1
    df['Sex'] = df['Sex'].map({'Female': 0, 'Male': 1})
    
    # 确保'Sex'列为int类型
    df['Sex'] = df['Sex'].astype(int)
    
    return df
```

```python
df = process_sex_column(df)
df.to_csv('processed_file.csv', index=False)
```
