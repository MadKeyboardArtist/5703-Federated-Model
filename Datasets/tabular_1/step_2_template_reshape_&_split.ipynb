```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
file_path = "diabetes_012_ready_to_model.csv"
```

```python
df_raw = pd.read_csv(file_path)
```

```python
df_raw.info()
```

1. reshape to meet the template

```python
# extract the 8 features in the template
import json
template_json = "../tabular_template_building/tabular_feature_template.json" 

with open(template_json, "r") as f:
  templete = json.load(f)
```

```python
t_features = templete["features"]
t_mask = templete["masks"]

t_features
```

```python
df_8_features = df_raw.copy()

df_8_features = df_8_features[t_features]
df_8_features.info()
```

```python
for col in t_mask:
    df_8_features[col] = 1

df_8_features.head()
```

```python
df_label = df_raw["Diabetes_012"]
df_tabular_ready = pd.concat(
    [df_8_features, df_label], 
    axis=1
    )

df_tabular_ready.head()
```

```python
df_tabular_ready.info()
```

2. record feature_means, feature_vars and sample_count for global z-score norm in training

```python
# drop the label column and any mask columns when computing feature stats
label_col = "Diabetes_012"
feature_cols = [c for c in df_tabular_ready.columns if not c.endswith("_mask") and c != label_col]

# compute per-feature mean and variance
feature_means = df_tabular_ready[feature_cols].mean().to_dict()
feature_vars = df_tabular_ready[feature_cols].var(ddof=0).to_dict()   # population variance

# build the JSON dict
site_stats = {
    "site_name": "tabular_1",
    "n_samples": int(len(df_tabular_ready)),
    "feature_means": feature_means,
    "feature_vars": feature_vars
}
site_stats
```

```python
import json

with open("tabular_1_stats.json", "w", encoding="utf-8") as f:
    json.dump(site_stats, f, indent=2)
```

3. train-val-test split

```python
# train-val-test split
# train + val vs test
tabular_train_val, tabular_test = train_test_split(
    df_tabular_ready, 
    test_size    = 0.2, 
    random_state = 42, 
    stratify = df_tabular_ready["Diabetes_012"]
    )

# train vs val
tabular_train, tabular_val = train_test_split(
    tabular_train_val, 
    test_size    = 0.25, 
    random_state = 42, 
    stratify = tabular_train_val["Diabetes_012"]
    )
```

```python
# save to 3 CSVs
tabular_train.to_csv("diabetes_012_train.csv", index = False)
tabular_val.to_csv  ("diabetes_012_val.csv"  , index = False)
tabular_test.to_csv ("diabetes_012_test.csv" , index = False)

print("Train/Val/Test saved successfully.")
```
