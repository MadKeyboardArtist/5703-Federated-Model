```python


import json

DATASET_FOLDER = "Datasets"

# ---- tabular_rex settings ----
name      = "tabular_rex"
modality  = "tabular"
label_col = "label"          
n_classes = 3

train = "tabular_rex_train.csv"
val   = "tabular_rex_val.csv"
test  = "tabular_rex_test.csv"

clean_dataset_train = f"{DATASET_FOLDER}/{name}/{train}"
clean_dataset_val   = f"{DATASET_FOLDER}/{name}/{val}"
clean_dataset_test  = f"{DATASET_FOLDER}/{name}/{test}"

info = {
    "name": name,
    "modality": modality,
    "label_col": label_col,
    "n_classes": n_classes,
    "clean_dataset_train": clean_dataset_train,
    "clean_dataset_val":   clean_dataset_val,
    "clean_dataset_test":  clean_dataset_test
}

out_file = f"{name}_info.json"   
with open(out_file, "w", encoding="utf-8") as f:
    json.dump(info, f, indent=4)
print("Wrote:", out_file)
```
